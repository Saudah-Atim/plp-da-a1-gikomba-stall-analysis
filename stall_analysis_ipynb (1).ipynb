##  Daily Sales Analysis
The total, average, highest and lowest daily sales for the 14 days. I will also identify the best and worst sales days and their respective weeks.

```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,

2500, 2200, 3100, 2400, 3900, 5600, 4700]  # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2



# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg

stock = {

"Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},

"Onions": {"qty": 40, "reorder": 20, "cost": 70, "price": 100},

"Sukuma": {"qty": 9, "reorder": 15, "cost": 30, "price": 50},

"Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},

"Avocado": {"qty": 12, "reorder": 10, "cost": 80, "price": 140},

}

# Calculate the total sales for the 14 days
total_sales = sum(daily_sales)

# Calculate avreage daily sales
average_sales = total_sales / len(daily_sales)

# Find the highest and lowest sales for the full 14 days
highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

#Let us look for the position(indexes which is the postion of an item in the list)
# We shall use the indexes to identify the coresponding days becasue both lsts have matching position
best_index = daily_sales.index(highest_sales)
worst_index = daily_sales.index(lowest_sales)

#Use the indexes to find the best and worst days
best_day = days[best_index]
worst_day = days[worst_index]

# we shall determine the week number using the indexes
# 0-6 belong to week 1 while 7-13 belong to week 2
best_week = (best_index // 7) + 1
worst_week = (worst_index // 7) + 1

#Display results
print("Daily Sales Analysis")
print("----------------------------")
print(f"Total Sales: KES {total_sales:,}")
print(f"Average Daily Sales: KES {average_sales:,.2f}")
print(f"Highest Sales: KES {highest_sales:,}")
print(f"Lowest Sales: KES {lowest_sales:,}")
print(f"Best Day: {best_day}, Week {best_week}")
print(f"Worst Day: {worst_day}, Week {worst_week}")
```

The week total sales and week-on-week growth percentage
Calculate the total sales for both week 1 and 2. We slice week 1 and 2 and use tuple method to return them both and calculate the growth percenatge

```python
def week_totals(sales):
  """ Calculate the total sales of week 1 and week 2.
  Return both weekly totals as a tuple """

  # separate the sales into two weeks
  week1 = sales[:7]
  week2 = sales[7:]

  # calculate the total sales for each week
  week1_total = sum(week1)
  week2_total = sum(week2)

  # return both weeks as a tuple
  return week1_total, week2_total

# Call the function globally to assign values
week1_total, week2_total = week_totals(daily_sales)

# Calculate the week-on-week growth percentage globally
growth_percentage = ((week2_total - week1_total) / week1_total) * 100

# Display the weekly sales results
print("WEEKLY SALES ANALYSIS")
print("----------------------------")
print(f"Week 1 Total: KES {week1_total:,}")
print(f"Week 2 Total: KES {week2_total:,}")
print(f"Week-on-Week Growth: {growth_percentage:.2f}%")
```

Classifying sales days using Strong, Weak and Normal using Average sales. Using a dictionary to count how many days fall under each category

```python
def classify_day(amount, average):
  """
  Classify daily sales as Strong, Normal or Weak.
  """
  # Strong means sales are more than 20% above average
  if amount > average * 1.20:
    return "Strong"
  # Weak means sales are more than 20% below average
  elif amount < average * 0.80:
    return "Weak"
  # Sales that are between Strong and Weak
  else:
    return "Normal"

# Initialize daily category counter dictionary
days_count = {
  "Strong": 0,
  "Weak": 0,
  "Normal": 0
}

# Loop through every day's sales
for amount in daily_sales:
  # Classify each day's sales using the average
  category = classify_day(amount, average_sales)
  # The count for the matching categories
  days_count[category] += 1

# Display the results
print("Sales classification")
print("----------------------------")
print(f"Strong: {days_count['Strong']}")
print(f"Weak: {days_count['Weak']}")
print(f"Normal: {days_count['Normal']}")
```

Average sales for weekdays Monday to Friday  Vs Avrage sales for Weekends Saturday and Sunday. i will show both of them in percentage **bold text**


```python
# create empty lists to store weekday and weekend sales
weekend_sales = []
weekday_sales = []

# loop through the positions in the daily_sales list
for i in range(len(daily_sales)):
  # get the day name and its matching sales amount
  day = days[i]
  amount = daily_sales[i]

  # check if the day is monday to friday
  if day in ["Mon", "Tue", "Wed", "Thu", "Fri"]:
    weekday_sales.append(amount)
  elif day in ["Sat", "Sun"]:
    weekend_sales.append(amount)

# calculate the average sales for weekdays and weekends
weekday_average = sum(weekday_sales) / len(weekday_sales)
weekend_average = sum(weekend_sales) / len(weekend_sales)

# calculate the percentage difference
percentage_difference = ((weekend_average - weekday_average) / weekday_average) * 100

# display the results
print("weekday_sales VS weekend_sales")
print("----------------------------")
print(f"Weekday Average: KES {weekday_average:,.2f}")
print(f"Weekend Average: KES {weekend_average:,.2f}")
print(f"Percentage Difference: {percentage_difference:.2f}%")
```

Calculate the gross margin percentage for each product using its selling price and cost price. Identifying which products need restocking

```python
# function to identify products that need restocking
def restock_products(stock):
  """ Calculate the gross margin percentage for each product.
  """
  # create an empty list to store products that need restocking
  reorder_list = []
  #go through each product and its details in the stock dictionary
  for product, details in stock.items():

    # calculate the gross margin percentage
    gross_margin = ((details["price"] - details["cost"]) / details["cost"]) * 100
    # check if the current quantity is at or below the order level
    if details["qty"] <= details["reorder"]:
      # add the product to the reorder list
      reorder_list.append(product)
  return reorder_list
for product, details in stock.items():
    margin = ((details["price"] - details["cost"]) / details["price"]) * 100
    print(f"{product}: {margin:.2f}%")


```

```python
# call the function to get the products that need restocking
reorder_list = restock_products(stock)

# display the results
print("Products that need restocking")
print("----------------------------")
for product in reorder_list:
  print(product)
```
